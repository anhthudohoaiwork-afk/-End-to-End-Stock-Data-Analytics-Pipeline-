# TASK 1: CHANGE BY DATE

```python
task_1 = """

-- Previously_close_price

WITH cte AS(
SELECT Date, Ticker, Close,
LAG(Close) OVER(PARTITION BY Ticker ORDER BY Date) as prev_price
FROM stock_prices_raw)

-- Task 2: Top 5 ngày peak theo daily change/ pct_change

SELECT Date, Ticker, Close, prev_price,
ROUND((Close - prev_price),4) as top_5_daily_change,
ROUND(((Close - prev_price)*100.0/NULLIF(prev_price,0)),4) as top_5_pct_change
FROM cte
WHERE Ticker = 'VCB'
-- ORDER BY top_5_daily_change DESC
ORDER BY top_5_pct_change DESC
LIMIT 5;

"""

cur.execute(task_1)

rows = cur.fetchall()
for row in rows:
     print(row)
```

# TASK 2: TOP 5 NGAY PEAK THEO DAILY CHANGE/ PCT CHANGE

```python
task_2 = """

WITH cte AS(
SELECT Date, Ticker, Close,
LAG(Close) OVER(PARTITION BY Ticker ORDER BY Date) as prev_price
FROM stock_prices_raw),

changes as(
SELECT Date, Ticker, Close, prev_price,
ROUND((Close - prev_price),4) as daily_change,
ROUND(((Close - prev_price)*100.0/NULLIF(prev_price,0)),4) as pct_change
FROM cte),

rank as(
SELECT Date, Ticker, Close, daily_change, pct_change,
ROW_NUMBER () OVER(PARTITION BY Ticker ORDER BY daily_change DESC) AS row_up_daily,
ROW_NUMBER() OVER(PARTITION BY Ticker ORDER BY daily_change ASC) AS row_down_daily,
ROW_NUMBER () OVER (PARTITION BY Ticker ORDER BY pct_change DESC) as row_up_pct,
ROW_NUMBER () OVER (PARTITION BY Ticker ORDER BY pct_change ASC) as row_down_pct
FROM changes)

SELECT Date, Ticker, Close, daily_change
FROM rank
WHERE  row_up_daily <=5

"""

cur.execute(task_2)

rows = cur.fetchall()
for row in rows:
     print(row)
```

TASK 5: PCT CHANGE OF TICKER BY QUARTER

```python
task_5 = """
-- Format date thanh int

WITH cte6 as(
SELECT Date, Ticker,
CAST(strftime('%Y', Date) AS INTEGER) AS Year,
((CAST(strftime('%m', Date) AS INTEGER) -1)/3) + 1 AS Quarter
FROM stock_prices_raw),

-- Sort last quarter close day

cte7 as(
SELECT Ticker, Year, Quarter,
MAX(Date) as last_close_day
FROM cte6
GROUP BY Ticker, Year, Quarter),

-- Lay quarter close price

cte8 as(
SELECT cte7.Ticker, cte7.Year, cte7.Quarter,
cte7.last_close_day, stock_prices_raw.Close as quarter_close
FROM cte7
JOIN stock_prices_raw
ON cte7.last_close_day = stock_prices_raw.Date
AND cte7.Ticker = stock_prices_raw.Ticker),

-- Lay prev quarter close price

cte9 as(
SELECT Ticker, Year, Quarter, quarter_close,
LAG(quarter_close) OVER(PARTITION BY Ticker ORDER BY Year, Quarter) as prev_quarter_close
FROM cte8)

SELECT Ticker, Year, Quarter,
ROUND(((quarter_close - prev_quarter_close)*100.0/NULLIF(prev_quarter_close,0)),4) as pct_change_by_quarter,
CASE
WHEN Quarter = 1 THEN 'Q1-' || Year
WHEN Quarter = 2 THEN 'Q2-' || Year
WHEN Quarter = 3 THEN 'Q3-' || Year
ELSE 'Q4-' || Year
END as quarter_label
FROM cte9
ORDER BY Ticker, Year, Quarter;

"""

# Run query
cur.execute(task_5)


# Lay ket qua
rows = cur.fetchall()
for row in rows:
     print(row)
```

TASK 6: AVG PCT CHANGE OF TICKER BY QUARTER

```python
task_6 = """


-- Lọc quý

WITH cte1 AS(
SELECT Date, Ticker, Close,
CAST(strftime('%Y', Date) AS INTEGER) AS year,
((CAST(strftime('%m', Date) AS INTEGER) -1)/3) +1 AS quarter
FROM stock_prices_raw),

-- Lấy last close day của mỗi quý

cte2 AS(
SELECT year, quarter, ticker,
MAX(Date) as last_close_day
FROM cte1
GROUP BY year, quarter, ticker),

-- Lấy last close day của mỗi quý

 cte3 AS(
SELECT cte2.year, cte2.quarter, cte2.ticker, cte2.last_close_day, stock_prices_raw. close as close_quarter
FROM cte2
JOIN stock_prices_raw
ON cte2.last_close_day = stock_prices_raw.date
AND cte2.ticker = stock_prices_raw.ticker
ORDER BY year ASC, quarter ASC),

-- Lấy prev quarter close price

cte4 as(
SELECT Year, quarter, ticker, last_close_day, close_quarter,
LAG(close_quarter) OVER(PARTITION BY Ticker ORDER BY Year, quarter) as prev_close_quarter
FROM cte3),

-- Tính phần trăm thay đổi theo quý

cte5 as(
SELECT Year, quarter, ticker, last_close_day,
ROUND(((close_quarter - prev_close_quarter)*100.0/ NULLIF(prev_close_quarter,0)),2) as pct_c_quarterly
FROM cte4
WHERE prev_close_quarter IS NOT NULL )

-- Tính phần trăm thay đổi trung bình của mỗi quý, quarter label

SELECT Year, quarter, ticker,
CASE
WHEN quarter = 1 THEN 'Q1-' || Year
WHEN quarter = 2 THEN 'Q2-' || Year
WHEN quarter = 3 THEN 'Q3-' || Year
ELSE 'Q4-' || Year
END as quarter_label,
ROUND(AVG(pct_c_quarterly),2) as avg_pct_c_by_quarter
FROM cte5
GROUP BY 1,2,3
ORDER BY Year ASC, quarter ASC, ticker

"""


# Run query
cur.execute(task_6)


# Lay ket qua
rows = cur.fetchall()
for row in rows:
     print(row)
```

TASK 10: So sánh pct change của Raw và Hist, check độ consistency của data

```python
task_10 = """

-- Prev close cua file raw

WITH cte as(
SELECT Date, Ticker, Close,
LAG(Close) OVER(PARTITION BY Ticker ORDER BY Date) as Prev_close
FROM stock_prices_raw),

-- Pct cua file raw

cte1 AS(
SELECT Date, Ticker,
(Close - Prev_close)*100.0/NULLIF(Prev_close,0) as pct_change
FROM cte),

-- Difference change cua raw vs hist

cte2 as(
SELECT
stock_prices_raw.Ticker,
stock_prices_raw.Date as time_raw,
stock_prices_hist.Date as time_hist,
ROUND(cte1.pct_change,4) as pct_change_raw,
ROUND(stock_prices_hist.pct_change,4) as pct_change_hist,
ROUND((cte1.pct_change - stock_prices_hist.pct_change),4) as diff_change
FROM stock_prices_raw
JOIN stock_prices_hist ON stock_prices_raw.date = stock_prices_hist.date
AND stock_prices_raw.ticker = stock_prices_hist.ticker
JOIN cte1 ON stock_prices_raw.date = cte1.date
AND stock_prices_raw.ticker = cte1.ticker),


-- Row count, diff change by ticker

cte3 as(
SELECT ticker,time_raw, time_hist, pct_change_raw,
pct_change_hist, diff_change,
COUNT(*) OVER (PARTITION BY ticker) as row_count,
ROW_NUMBER () OVER(PARTITION BY Ticker ORDER BY Diff_change ASC) AS row_num
FROM cte2)

-- Threshold Median of ticker

SELECT ticker,
AVG(Diff_change) as median
FROM cte3
WHERE row_num IN (row_count/2, (row_count+1)/2)
GROUP BY Ticker;

"""

cur.execute(task_10)

rows = cur.fetchall()
for row in rows:
    print(row)


```

TASK 11: OUTLIER DAY CHECK BY PCT CHANGE

```python
task_11 = """

-- Pct change

WITH cte1 as(
SELECT Date, Ticker, Close,
ROUND(((Close - LAG(Close) OVER(PARTITION BY Ticker ORDER BY Date))*100.0/NULLIF(LAG(Close) OVER(PARTITION BY Ticker ORDER BY Date),0)),4)
as pct_change
FROM stock_prices_raw),

-- Mean, stddev

cte2 AS(
SELECT Ticker,
ROUND(AVG(pct_change),4) AS mean_pct_change,
ROUND((SQRT(AVG(pct_change*pct_change) - (AVG(pct_change)*AVG(pct_change)))),4) as stddev_pct_change
FROM cte1
GROUP BY Ticker)

-- Outlier check

SELECT cte1.Date, cte1.Ticker, cte1.pct_change,
cte2.mean_pct_change, cte2.stddev_pct_change,
ROUND((cte2.mean_pct_change + 2*cte2.stddev_pct_change),4) as threshold1,
ROUND((cte2.mean_pct_change - 2*cte2.stddev_pct_change),4) as threshold2,
CASE
WHEN cte1.pct_change > ROUND((cte2.mean_pct_change + 2*cte2.stddev_pct_change),4) THEN 'Positive Outlier'
WHEN cte1.pct_change < ROUND((cte2.mean_pct_change - 2*cte2.stddev_pct_change),4) THEN 'Negative Outlier'
ELSE 'Normal'
END as outlier_check
FROM cte1
JOIN cte2 ON cte1.Ticker = cte2.Ticker


"""

cur.execute(task_11)

column = ["Date", "Ticker", "Pct_change", "Mean_pct_change", "stddev_pct_change",
          "threshold1", "threshold2", "Outlier_check"]
print(column)

rows = cur.fetchall()
for row in rows:
     print(row)







```

TASK 12: ROLLING AVERAGE

```python
task_12 = """

WITH cte as(
SELECT Date, Ticker, Close,
ROUND(((Close - LAG(Close) OVER(PARTITION BY Ticker ORDER BY Date))*100.0/NULLIF(LAG(Close) OVER(PARTITION BY Ticker ORDER BY Date),0)),3) AS pct_change
FROM stock_prices_raw),

short_term as(
SELECT Date, Ticker,
ROUND((AVG(pct_change) OVER (PARTITION BY Ticker ORDER BY Date
ROWS BETWEEN 6 PRECEDING AND CURRENT ROW)),3) as rolling_7days
FROM cte),

long_term as (
SELECT Date, Ticker,
ROUND((AVG(pct_change) OVER(PARTITION BY ticker ORDER BY Date ROWS BETWEEN 29 PRECEDING AND CURRENT ROW)),3) AS rolling_30days
FROM cte)


SELECT cte.Date, cte.Ticker as Ticker, short_term.rolling_7days as rolling_avg_short,
long_term.rolling_30days as rolling_avg_long
FROM cte
JOIN short_term ON cte.Date = short_term.Date
AND cte.Ticker = short_term.Ticker
JOIN long_term ON cte.Date = long_term.Date
AND cte.Ticker = long_term.Ticker

"""

cur.execute(task_12)

column = ["Date", "Ticker", "rolling_7days", "rolling_30days"]

print(column)

rows = cur.fetchall()
for row in rows:
     print(row)
```

TASK 13: RANKING BY DAY

```python
task_13 = """


WITH cte as(
SELECT Date, Ticker, Close,
ROUND(((Close - (LAG(Close) OVER(PARTITION BY Ticker ORDER BY Date)))*100.0/NULLIF(LAG(Close) OVER(PARTITION BY Ticker ORDER BY Date),0)),3) as pct_change
FROM stock_prices_raw),

rank_group as (
SELECT Date, Ticker, pct_change,
RANK () OVER (PARTITION BY Ticker ORDER BY pct_change DESC) as rank_up,
RANK () OVER (PARTITION BY Ticker ORDER BY pct_change ASC) as rank_down
FROM cte)

-- Top 5 peak

SELECT Ticker, Date, pct_change, rank_up
FROM rank_group
WHERE rank_up <= 5

UNION ALL

-- Top 5 weak

SELECT Ticker, Date, pct_change, rank_down
FROM rank_group
WHERE rank_down <=5
ORDER BY Ticker


"""

cur.execute(task_13)

column = ["Ticker", "Date", "pct_change", "Rank_days"]

print(column)

rows = cur.fetchall()
for row in rows:

      print(row)



```

# 14. Build Pipeline (Scaling + Model)

```python
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score
```

```python
input = [("scale",StandardScaler()), ("model", LinearRegression())]

pipe=Pipeline(input)
```

```python
X1 = df_vcb[x_train].astype('float')
Y1 = df_vcb["close_price"]


X2 = df_vib[x_train].astype('float')
Y2 = df_vib["close_price"]


X3 = df_vic[x_train].astype('float')
Y3 = df_vic["close_price"]
```

```python
pipe.fit(X1,Y1) # Train VCB
```

```python
pipe.fit(X2,Y2) # Train VIB
```

```python
pipe.fit(X3,Y3) # Train VIC
```



> **MODEL**



```python
y1pipe=pipe.predict(X1) # Model VCB
```

```python
y2pipe=pipe.predict(X2) # Model VIB
```

```python
y3pipe=pipe.predict(X3) # Model VIC
```

```python
# VCB

mse1 = mean_squared_error(df_vcb["close_price"], y1pipe)
print("The VCB MSE of Close price and predicted value is:", mse1)

R2_score1 = r2_score(df_vcb["close_price"],y1pipe)
print("The VCB R-squared of Linear Regression model is:", R2_score1)
```

```python
# VIB

mse2 = mean_squared_error(df_vib["close_price"],y2pipe)
print("The VIB MSE of Close price and predicted value is:", mse2)

R2_score2 = r2_score(df_vib["close_price"], y2pipe)
print("The VIB R-squared of Linear Regression model is:", R2_score2)
```

```python
# VIC

mse3 = mean_squared_error(df_vic["close_price"],y3pipe)
print("The VIC MSE of Close price and predicted value is:", mse3)

R2_score3 = r2_score(df_vic["close_price"],y3pipe)
print("The VIC R-squared of Linear Regression model is:", R2_score3)
```
